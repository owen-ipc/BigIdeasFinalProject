# Police Arrest Data in past 2 years

## Information about the data set

**Publisher**   
The city of Pittsbugh

**What am I measuring**
I'm measuring how many pepole get arrested in each neighborhood in the past 2 years. This data makes since because less arrests means less crimes and less crimes means more peacfull neighborhood.

**Cleaning steps**
I cleaned my data set by looking at each neighborhood possible, and then removing the three neighborhoods that were not in the city or in the surrounding area. The three rows I dropped were either outside the city, state, or county. Then there was one name that I didn't really like it was central northside. I decided to combine that into one word central north side instead of north space side.

**Charts**
I have two charts. The first chart shows the worst neighborhoods, with the most arrests by city police officers. The second chart shows the best neighborhoods, with the least arrests by city police officers. You can see that the business district is the worst neighborhood, but that also goes with that neighborhood probably has one of the most visitors because it is very big. The best neighborhood that I found was Mt. Oliver Neighborhood it had very few arrests and I'm pretty sure it is a gated community.

```python
import pandas as pd

data = pd.read_csv("data.csv")

#This part of the code if me cleaning my data set

clean = data[data["INCIDENTNEIGHBORHOOD"].notna()].copy()
clean["INCIDENTNEIGHBORHOOD"] = clean["INCIDENTNEIGHBORHOOD"].replace("Central North Side", "Central Northside")
counts = clean["INCIDENTNEIGHBORHOOD"].value_counts().reset_index()
counts.columns = ["neighborhood", "arrests"]
outside = ["Outside City", "Outside State", "Outside County"]  #There where values that where outside the scope of the project
idx = counts[counts["neighborhood"].isin(outside)].index
counts = counts.drop(idx) 
counts = counts.reset_index(drop = True)     #Final data frame
print(counts.shape)
```
